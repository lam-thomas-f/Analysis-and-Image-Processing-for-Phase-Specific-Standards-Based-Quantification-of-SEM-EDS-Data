```python
#qt for gui figures, inline for embedded in notebook 
%matplotlib qt
%matplotlib inline 
import hyperspy.api as hs
import exspy
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
s = hs.load("Alm_0p078.rpl", stack=True)
s.set_signal_type("EDS_SEM")
s.metadata.General.title = 'Almandine Standard for Fe'
s
```

```python
s.axes_manager.gui()
#enter pixel, and -0.96 offset keV and 0.01 keV, scale is 0.078 microns
```

```python
sum_spec = s.sum()
sum_spec.set_elements(["Mg", "Al", "Si", "Ca", "Mn", "Fe"])
sum_spec.add_lines(["Mg_Ka", "Al_Ka", "Si_Ka", "Ca_Ka", "Mn_Ka", "Fe_Ka"])
sum_spec.plot(True)
sum_spec._plot.signal_plot.ax.set_xlim(0.2, 7)
```

```python
ss = sum_spec
x = ss.axes_manager.signal_axes[0].axis
y = ss.data

left_bg = (x >= 6.05) & (x <= 6.2)
right_bg = (x >= 6.65) & (x <= 6.8)
peak = (x >= 6.26) & (x <= 6.54)

x_bg = np.r_[x[left_bg], x[right_bg]]
y_bg = np.r_[y[left_bg], y[right_bg]]

coef = np.polyfit(x_bg, y_bg, 1)
baseline = np.polyval(coef, x)

y_corr = y - baseline
Fe_net = y_corr[peak].sum()

print("Fe_net_counts", f"{Fe_net:.3f}")
```

```python
df = pd.read_csv("15kV_Stds_Lib.csv")
df.iloc[0, 11] = Fe_net
df.to_csv("15kV_Stds_Lib.csv", index=False)
```
